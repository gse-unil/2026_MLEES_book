## Exercise 3: A pure function

```python
def mean(values):
    total = 0.0
    for v in values:
        total += v
    return total / len(values)

readings_celsius = [6.1, 7.4, 5.9, 8.2, 6.8]
print(round(mean(readings_celsius), 2))   # 6.88
```

```python
def is_warm(temp_celsius, threshold=10.0):
    if temp_celsius >= threshold:
        return True
    else:
        return False

print(is_warm(6.1))                     # False
print(is_warm(6.1, threshold=5.0))      # True
```

## Exercise 6: Number the readings with enumerate

```python
week_celsius = [6.1, 7.4, 5.9, 8.2, 11.3, 9.8, 6.8]

warmest_celsius = week_celsius[0]     # start from the first reading, not from 0.0:
warmest_day = 1                       # a week of frost would never beat a 0.0 starting point

for index, temp_celsius in enumerate(week_celsius):
    day = index + 1                   # enumerate counts from 0; days are counted from 1
    print(f"day {day}: {temp_celsius} degC")
    if temp_celsius > warmest_celsius:
        warmest_celsius = temp_celsius
        warmest_day = day

print(f"warmest: {warmest_celsius} degC on day {warmest_day}")     # 11.3 degC on day 5
```

## Exercise 7: while, break, and continue

```python
mixed_celsius = [6.1, None, 5.9, 9.5, 6.8]

first_hot = None          # defined first, in case no hot reading is found
i = 0
while i < len(mixed_celsius):
    value = mixed_celsius[i]
    i += 1                            # advance before any continue, or the loop never ends
    if value is None:
        continue
    if value > 9.0:
        first_hot = value
        break

print("first hot reading:", first_hot)      # 9.5
```

## Exercise 8: Default arguments and returning two values

```python
def summarise(values, scale=1.0):
    """Return the total and the mean of a list of numbers, after scaling."""
    total = 0.0
    for v in values:
        total += v * scale
    return total, total / len(values)
```

```python
readings_celsius = [6.1, 7.4, 5.9, 8.2, 6.8]
```

```python
total_celsius, mean_celsius = summarise(readings_celsius)
print(round(total_celsius, 2), round(mean_celsius, 2))     # 34.4 6.88
```

## Exercise 9: A variable number of arguments

```python
def warmest(*temps_celsius):
    """Return the highest value among any number of positional arguments."""
    best = temps_celsius[0]
    for t in temps_celsius:
        if t > best:
            best = t
    return best

print(warmest(6.1, 9.5, 5.9))                        # 9.5
readings_celsius = [6.1, 7.4, 5.9, 8.2, 6.8]
print(warmest(*readings_celsius))                     # 8.2
```

## Exercise 10: Fix the shared-default bug

```python
# out=[] is created once at definition and shared across calls, so results accumulate.
# the simplest fix is to drop the out parameter and build a fresh list inside.
def collect_above(values, threshold):
    """Return the values above a threshold."""
    out = []
    for v in values:
        if v > threshold:
            out.append(v)
    return out

first = collect_above([1.0, 5.0, 9.0], 4.0)
second = collect_above([2.0, 8.0], 4.0)
print(first)    # [5.0, 9.0]
print(second)   # [8.0], not contaminated
```

### Exercise 14: Spot the impure function

```python
def drop_last_impure(readings):
    readings.pop()
    return readings
```

```python
original = [6.1, 7.4, 5.9]
drop_last_impure(original)
print(original)                    # [6.1, 7.4] — the caller's list was modified
```

```python
def drop_last_pure(readings):
    """Return a copy of the list without its last element."""
    trimmed = readings.copy()
    trimmed.pop()
    return trimmed
```

```python
original = [6.1, 7.4, 5.9]
shorter = drop_last_pure(original)
print(original, shorter)           # [6.1, 7.4, 5.9] [6.1, 7.4]

# both bugs come from the same fact: mutable objects are passed by reference, so a function
# can reach back into data it does not own — including its own default argument
```

### Exercise 15: Reject impossible temperatures

```python
def to_kelvin(temp_celsius: float) -> float:
    """Return a temperature converted from degrees Celsius to kelvin."""
    if temp_celsius < -273.15:
        raise ValueError(f"{temp_celsius} °C is below absolute zero")
    return temp_celsius + 273.15

for value in [20.0, -300.0, 0.0]:
    try:
        print(round(to_kelvin(value), 2))
    except ValueError as err:
        print("skipped:", err)
```

## Exercise 16: A year of station data, continued

In subchapter 1.1 you fetched this file, worked out its structure, and converted a single
record to °C. You did not yet have the tools to handle all of it at once — now you do.

The goal: turn a year of daily maximum temperatures into a monthly summary, and write that
summary out as a new file. The input file has the same three columns as in Exercise 9 of the
previous subchapter: the station code, the day as `dd.mm.yy`, and the daily maximum temperature
in degrees Fahrenheit. Your monthly summary will have its own columns: `month`, `n_obs`
(the number of observations), and `mean_temp_celsius`.

```python
# Pre-supplied: download the data file and cache it locally (see subchapter 1.1).
import pooch

data_file = pooch.retrieve(
    url="https://raw.githubusercontent.com/gse-unil/2026_MLEES_book/main/data/part-I/station_iib_daily_max_temp_2022.csv",
    known_hash="sha256:034755fb289b4e157e5d029995481786a359eb25a80df62c09ee83d063013144",
    fname="station_iib_daily_max_temp_2022.csv",
    path=pooch.os_cache("mlees"),
)
print("data cached at:", data_file)
```

**Step 1.** The conversion, written once as a function.

```python
def fahrenheit_to_celsius(temp_fahrenheit):
    """Convert a temperature from degrees Fahrenheit to degrees Celsius."""
    return (temp_fahrenheit - 32.0) * 5.0 / 9.0


print(fahrenheit_to_celsius(32.0))    # freezing point of water: 0.0
print(fahrenheit_to_celsius(212.0))   # boiling point of water: 100.0
```

**Step 2.** Loop over the records, pulling out the month and the temperature in °C.

```python
with open(data_file, "r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()
```

```python
# 16 days in this file have an empty temperature field -- a real sensor gap. float("")
# raises ValueError, so the loop has to decide what to do with them; here it skips them
# and counts them, and step 7 turns that into a general guard.
n_records = 0
n_gaps = 0
for line in lines[1:]:                      # skip the header
    fields = line.strip().split(",")
    day = fields[1]                         # dd.mm.yy
    month = day.split(".")[1]
    if fields[2] == "":
        n_gaps += 1
        continue
    temp_celsius = fahrenheit_to_celsius(float(fields[2]))
    if n_records < 3:
        print(f"month {month}: {temp_celsius:.1f} °C")
    n_records += 1
```

```python
print("records processed:", n_records, "| days with no temperature:", n_gaps)
```

**Step 3.** Count the observations per month in a dictionary that grows as it goes.

```python
counts = {}
for line in lines[1:]:
    fields = line.strip().split(",")
    if fields[2] == "":          # a day with no reading is not an observation
        continue
    month = fields[1].split(".")[1]
    counts[month] = counts.get(month, 0) + 1

print(counts)
```

**Step 4.** Accumulate a running total per month, then turn each total into a mean.

```python
counts = {}
totals = {}
for line in lines[1:]:
    fields = line.strip().split(",")
    if fields[2] == "":
        continue
    month = fields[1].split(".")[1]
    temp_celsius = fahrenheit_to_celsius(float(fields[2]))
    counts[month] = counts.get(month, 0) + 1
    totals[month] = totals.get(month, 0.0) + temp_celsius
```

```python
means = {}
for month in totals:
    means[month] = totals[month] / counts[month]
```

```python
for month in means:
    print(month, round(means[month], 2))
```

**Step 5.** The warmest month, found by keeping the best value seen so far.

```python
warmest_month = None
warmest_mean = None
for month in means:
    if warmest_mean is None or means[month] > warmest_mean:
        warmest_month = month
        warmest_mean = means[month]

print(f"warmest month: {warmest_month} at {warmest_mean:.2f} °C")
```

**Step 6.** Write the summary out, then read it back to check it says what you meant.

```python
from pathlib import Path

Path("_files").mkdir(exist_ok=True)
out_path = Path("_files") / "monthly_means.csv"

with open(out_path, "w", encoding="utf-8") as fhandler:
    fhandler.write("month,n_obs,mean_temp_celsius\n")
    for month in means:
        fhandler.write(f"{month},{counts[month]},{means[month]:.2f}\n")

with open(out_path, "r", encoding="utf-8") as fhandler:
    print(fhandler.read(), end="")
```

### Part III: functions

**Q9) Write a function that converts a planetary mass to Earth masses.**

Jupiter is about 1898 x 10^24 kg, which is about 318 Earth masses:

$$M_{\text{Jupiter}} \approx 1898 \times 10^{24}\,\text{kg} \approx 318\,M_{\text{Earth}}$$

Call it `to_earth_masses(mass_1e24kg)` and check it against Jupiter — you should get
317.92294807370183.

```python
def to_earth_masses(mass_1e24kg):
    return mass_1e24kg / planet_masses_1e24kg["earth"]

print(to_earth_masses(planet_masses_1e24kg["jupiter"]))   # 317.92294807370183
```

**Q10) Now write a single function that converts to either Earth masses or Jupiter
masses, depending on a keyword argument.**

Call it `to_planet_masses(mass_1e24kg, reference="jupiter")`. Converting Jupiter's own mass should
give 317.923 with `reference="earth"` and exactly 1.0 with `reference="jupiter"`.

A default argument is what turns two near-identical functions into one, and 1.0 is the check that
tells you the conversion is not quietly off by a factor somewhere.

```python
def to_planet_masses(mass_1e24kg, reference="jupiter"):
    return mass_1e24kg / planet_masses_1e24kg[reference]

print(to_planet_masses(planet_masses_1e24kg["jupiter"], "earth"))   # 317.92294807370183
print(to_planet_masses(planet_masses_1e24kg["jupiter"]))            # 1.0, the default reference
```

**Q11) Write a function that takes one argument and returns two.**

`mass_in_two_units(mass_1e24kg, reference)` should return the mass in Earth masses *and* in masses
of whatever reference planet you pass. Test it on Jupiter with `reference="mars"`.

Hint: `return a, b` returns a tuple, and `first, second = f(...)` unpacks it — both from 1.2's
lecture.

```python
def mass_in_two_units(mass_1e24kg, reference):
    in_earth_masses = mass_1e24kg / planet_masses_1e24kg["earth"]
    in_reference_masses = mass_1e24kg / planet_masses_1e24kg[reference]
    return in_earth_masses, in_reference_masses

in_earth, in_mars = mass_in_two_units(planet_masses_1e24kg["jupiter"], "mars")
print(in_earth, in_mars)      # 317.92294807370183 2956.386292834891
```

**Bonus.** Convert Neptune's mass to Jupiter masses with your Q10 function, then feed
that result back through Q11. Do you get the original value back? If not, work out which of the two
functions is being given a number in the wrong unit — this is the mistake the `_1e24kg` suffix
exists to prevent.

```python
neptune_in_jupiter_masses = to_planet_masses(planet_masses_1e24kg["neptune"])
print(neptune_in_jupiter_masses)                                  # 0.0537...

# feeding that straight back in does NOT recover the original
print(mass_in_two_units(neptune_in_jupiter_masses, "mars"))       # (0.009, 0.0837)
print(mass_in_two_units(planet_masses_1e24kg["neptune"], "mars")) # (17.09, 158.88)

# Q10 returned a number in Jupiter masses; Q11 expects one in 1e24 kg. The two functions
# disagree about the unit of their input, and nothing in the code says so -- which is why
# the suffix belongs in the variable name rather than in a comment.
```

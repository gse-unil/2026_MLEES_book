## Exercise 1: Build a catalogue, look up safely

You are given station records as a list of tuples `(code, name, elevation_m)`:

```python
[("JFJ", "Jungfraujoch", 3571), ("BAS", "Basel-Binningen", 316), ("LUG", "Lugano", 273)]
```

Build a dict mapping each `code` to a dict with keys `name` and `elevation_m`. Then look up code `"ZRH"` with `.get`, returning `"unknown"` if it is absent.

:::{admonition} Solution
:class: note dropdown
```python
records = [("JFJ", "Jungfraujoch", 3571), ("BAS", "Basel-Binningen", 316), ("LUG", "Lugano", 273)]
stations = {}
for code, name, elevation_m in records:
    stations[code] = {"name": name, "elevation_m": elevation_m}
print(stations["JFJ"])
print(stations.get("ZRH", "unknown"))
```

## Exercise 2: Classify and count

Given `readings_celsius = [-2.3, 6.1, 9.4, 15.0, 22.0, -0.5]`, classify each value as "freezing" (< 0), "cold" (< 10), "mild" (< 20), or "warm", and count how many fall in each category. Return the counts as a dict.

:::{admonition} Solution
:class: note dropdown
```python
readings_celsius = [-2.3, 6.1, 9.4, 15.0, 22.0, -0.5]
counts = {"freezing": 0, "cold": 0, "mild": 0, "warm": 0}
for t in readings_celsius:
    if t < 0.0:
        counts["freezing"] += 1
    elif t < 10.0:
        counts["cold"] += 1
    elif t < 20.0:
        counts["mild"] += 1
    else:
        counts["warm"] += 1
print(counts)
```

## Exercise 3: Warmest day by index

Given `dates = ["Mon", "Tue", "Wed", "Thu", "Fri"]` and `readings_celsius = [6.1, 7.4, 5.9, 8.2, 6.8]`, use `enumerate` to find the index of the warmest reading, then print the matching date and temperature.

:::{admonition} Solution
:class: note dropdown
```python
dates = ["Mon", "Tue", "Wed", "Thu", "Fri"]
readings_celsius = [6.1, 7.4, 5.9, 8.2, 6.8]
warmest_i = 0
for i, temp in enumerate(readings_celsius):
    if temp > readings_celsius[warmest_i]:
        warmest_i = i
print(dates[warmest_i], readings_celsius[warmest_i])
```

## Exercise 4: A pure function

Write a function `mean(values)` (no type hints) that returns the average of a list of numbers. Test it on `[6.1, 7.4, 5.9, 8.2, 6.8]`, printing the result rounded to two decimals.

:::{admonition} Solution
:class: note dropdown
```python
def mean(values):
    return sum(values) / len(values)

readings_celsius = [6.1, 7.4, 5.9, 8.2, 6.8]
print(round(mean(readings_celsius), 2))
```

## Exercise 5: A function with a default argument

Write a function `is_warm(temp_celsius, threshold=10.0)` (no type hints) that returns `True` when the temperature is at or above the threshold and `False` otherwise, using `if`/`else`. Test it on 6.1 °C with the default threshold, and again on 6.1 °C with a threshold of 5.0 °C passed by keyword.

:::{admonition} Solution
:class: note dropdown
```python
def is_warm(temp_celsius, threshold=10.0):
    if temp_celsius >= threshold:
        return True
    else:
        return False

print(is_warm(6.1))                 # False, since 6.1 < 10.0
print(is_warm(6.1, threshold=5.0))  # True, since 6.1 >= 5.0
```

## Exercise 6: Fix the shared-default bug

An assistant wrote this function; it is correct the first time but wrong on later calls:

```python
def collect_above(values, threshold, out=[]):
    for v in values:
        if v > threshold:
            out.append(v)
    return out
```

In one comment, explain the bug, then rewrite it so each call starts fresh. Demonstrate with two separate calls, assigning each result to its own variable.

:::{admonition} Solution
:class: note dropdown
```python
# out=[] is created once at definition and shared across calls, so results accumulate.
# the simplest fix is to drop the out parameter and build a fresh list inside.
def collect_above(values, threshold):
    out = []
    for v in values:
        if v > threshold:
            out.append(v)
    return out

first = collect_above([1.0, 5.0, 9.0], 4.0)
second = collect_above([2.0, 8.0], 4.0)
print(first)    # [5.0, 9.0]
print(second)   # [8.0], not contaminated
```

## Exercise 7: Means and a filter, with comprehensions

Recreate a `stations` dict where each code maps to a dict containing `readings_celsius`:

```python
{"JFJ": {"readings_celsius": [-2.3, -1.1, 0.4, 1.2, -0.8]},
 "BAS": {"readings_celsius": [6.1, 7.4, 5.9, 8.2, 6.8]},
 "LUG": {"readings_celsius": [9.4, 10.2, 8.8, 11.1, 9.9]}}
```

With a dict comprehension, build `means = {code: mean_temp}`. Then, with a list comprehension, build the list of codes whose mean is below 0 °C.

:::{admonition} Solution
:class: note dropdown
```python
stations = {
    "JFJ": {"readings_celsius": [-2.3, -1.1, 0.4, 1.2, -0.8]},
    "BAS": {"readings_celsius": [6.1, 7.4, 5.9, 8.2, 6.8]},
    "LUG": {"readings_celsius": [9.4, 10.2, 8.8, 11.1, 9.9]},
}
means = {code: sum(v["readings_celsius"]) / len(v["readings_celsius"])
         for code, v in stations.items()}
cold_codes = [code for code, m in means.items() if m < 0.0]
print({code: round(m, 2) for code, m in means.items()})
print(cold_codes)
```

## Exercise 2: Unit conversions

```python
temp_celsius = 18.5
temp_kelvin = temp_celsius + 273.15
temp_fahrenheit = temp_celsius * 9 / 5 + 32
print(temp_kelvin, temp_fahrenheit)          # 291.65 65.3

discharge_m3s = 4.7
area_km2 = 250.0
specific_discharge_mm_day = discharge_m3s / area_km2 * 86.4
print(round(specific_discharge_mm_day, 2))   # 1.62
```

## Exercise 3: Operator families and a quality flag

```python
reading_1_celsius = 18.2
reading_2_celsius = -300.0
reading_3_celsius = None

# 1. `and` requires both conditions to hold
is_plausible = reading_1_celsius > -273.15 and reading_1_celsius < 60.0
print(is_plausible)                                    # True

# 2. `not` negates a boolean
is_implausible = not (reading_2_celsius > -273.15)
print(is_implausible)                                  # True

# 3. `is` tests identity — the right way to check for None
print(reading_3_celsius is None)                       # True

# 4. a bool casts to int: True -> 1, False -> 0
n_valid = 0
n_valid += int(reading_1_celsius > -273.15)
n_valid += int(reading_2_celsius > -273.15)
print("n_valid:", n_valid)                             # 1
```

## Exercise 4: Parse a record

```python
record = "2024-01-15 , JUNGFRAUJOCH , -2.3 , degC"
fields = record.split(",")
date_str = fields[0].strip()
station_name = fields[1].strip().title()
value_celsius = float(fields[2].strip())
unit = fields[3].strip().lower()
print(date_str, station_name, value_celsius, unit)   # 2024-01-15 Jungfraujoch -2.3 degc
```

## Exercise 5: Indexing and slicing a record

```python
record = "ALO01.03.2022"

# 1. slicing: each field sits at a known position
station = record[:3]        # first three characters
day = record[3:5]
month = record[6:8]
year = record[-4:]          # the last four characters
print(station, day, month, year)      # ALO 01 03 2022

# 2. the same month, via a string method
date_part = record[3:]
month_again = date_part.split(".")[1]
print(month_again)                     # 03

# 3. slicing assumes fixed positions, so a two- or four-character station code shifts every
#    field and silently returns the wrong text. Splitting on a separator survives that,
#    so it is the safer choice whenever the field widths can vary.
```

## Exercise 6: Round-trip through a file

```python
from pathlib import Path
import math

readings_celsius = [-2.3, -1.1, 0.4, 1.2, -0.8, -3.1, -1.9]
path = Path("_files/my_week.txt")
path.parent.mkdir(exist_ok=True)

with path.open("w", encoding="utf-8") as fhandler:
    for value in readings_celsius:
        fhandler.write(f"{value}\n")

with path.open("r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

total = 0.0
for line in lines:
    total += float(line)
mean_celsius = total / len(lines)
print(round(mean_celsius, 2))          # -1.09

original_total = 0.0
for value in readings_celsius:
    original_total += value
original_mean = original_total / len(readings_celsius)
print(math.isclose(mean_celsius, original_mean))   # True
```

## Exercise 7: Round-trip through a file, with an append

```python
from pathlib import Path
import math

path = Path("_files/my_week.txt")
path.parent.mkdir(exist_ok=True)

# 1. "w" creates the file (or overwrites it)
with path.open("w", encoding="utf-8") as fhandler:
    fhandler.write("-2.3\n")
    fhandler.write("-1.1\n")
    fhandler.write("0.4\n")

# 2. "a" appends, keeping what is already there
with path.open("a", encoding="utf-8") as fhandler:
    fhandler.write("1.2\n")

# 3. read the file as one string
with path.open("r", encoding="utf-8") as fhandler:
    print(fhandler.read())

# 4. read it as separate lines and cast each one
with path.open("r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

print("number of lines:", len(lines))                  # 4
first = float(lines[0])
second = float(lines[1])
third = float(lines[2])
fourth = float(lines[3])
mean_celsius = (first + second + third + fourth) / len(lines)
print(f"mean = {mean_celsius:.2f} °C")                 # mean = -0.45 °C

# 5. the round trip should not change the mean
print(math.isclose(mean_celsius, (-2.3 + -1.1 + 0.4 + 1.2) / 4))   # True
```

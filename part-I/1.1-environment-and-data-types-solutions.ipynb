## Exercise 3: Operator families and a quality flag

```python
reading_1_celsius = 18.2
reading_2_celsius = -300.0
reading_3_celsius = None

# 1. `and` requires both conditions to hold
is_plausible = reading_1_celsius > -273.15 and reading_1_celsius < 60.0
print(is_plausible)                                    # True

# 2. `not` negates a boolean
is_implausible = not (reading_2_celsius > -273.15)
print(is_implausible)                                  # True

# 3. `is` tests identity — the right way to check for None
print(reading_3_celsius is None)                       # True

# 4. a bool casts to int: True -> 1, False -> 0
n_valid = 0
n_valid += int(reading_1_celsius > -273.15)
n_valid += int(reading_2_celsius > -273.15)
print("n_valid:", n_valid)                             # 1
```

## Exercise 5: Indexing and slicing a record

```python
record = "ALO01.03.2022"

# 1. slicing: each field sits at a known position
station = record[:3]        # first three characters
day = record[3:5]
month = record[6:8]
year = record[-4:]          # the last four characters
print(station, day, month, year)      # ALO 01 03 2022

# 2. the same month, via a string method
date_part = record[3:]
month_again = date_part.split(".")[1]
print(month_again)                     # 03

# 3. slicing assumes fixed positions, so a two- or four-character station code shifts every
#    field and silently returns the wrong text. Splitting on a separator survives that,
#    so it is the safer choice whenever the field widths can vary.
```

## Exercise 6: Formatting π to the precision you choose

Nobody remembers π beyond a few digits, but you do not need to: `math.pi` holds the full value, and an f-string format specifier prints it to whatever precision you choose.

1. Write the sentence `"This is my first I/O exercise."` to `_files/pi_exercise.txt` with mode `"w"`.
2. Reopen the file with mode `"a"` and append three more lines: one built with an f-string that reports `math.pi` to four decimal places, and two lines of your own choosing.
3. Read the file back with `.readlines()` and print its full contents.
4. Change the format specifier to two decimals instead of four, rerun, and confirm the printed value now rounds to `3.14`.

```python
import math
from pathlib import Path

Path("_files").mkdir(exist_ok=True)
path = Path("_files") / "pi_exercise.txt"

# 1. write the first line with mode "w", which creates the file or truncates it
with open(path, "w", encoding="utf-8") as fhandler:
    fhandler.write("This is my first I/O exercise.\n")

# 2. reopen with mode "a", which appends instead of truncating
with open(path, "a", encoding="utf-8") as fhandler:
    fhandler.write(f"Pi to four decimals is {math.pi:.4f}.\n")
    fhandler.write("The first line was written in mode 'w', the rest in mode 'a'.\n")
    fhandler.write("Opening in 'w' a second time would have thrown the first line away.\n")

# 3. read it back
with open(path, "r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

print(f"{len(lines)} lines:")
for line in lines:
    print(line, end="")   # each line already ends in a newline

# 4. two decimals instead of four -- 3.14159... rounds to 3.14
print()
print(f"four decimals: {math.pi:.4f}")
print(f"two decimals:  {math.pi:.2f}")
```

## Exercise 7: Round-trip through a file

```python
from pathlib import Path
import math

path = Path("_files/my_week.txt")
path.parent.mkdir(exist_ok=True)

with path.open("w", encoding="utf-8") as fhandler:
    fhandler.write("-2.3\n")
    fhandler.write("-1.1\n")
    fhandler.write("0.4\n")
    fhandler.write("1.2\n")
    fhandler.write("-0.8\n")
    fhandler.write("-3.1\n")
    fhandler.write("-1.9\n")

with path.open("r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

first = float(lines[0])
second = float(lines[1])
third = float(lines[2])
fourth = float(lines[3])
fifth = float(lines[4])
sixth = float(lines[5])
seventh = float(lines[6])
total = first + second + third + fourth + fifth + sixth + seventh
mean_celsius = total / len(lines)
print(round(mean_celsius, 2))          # -1.09

original_total = -2.3 + -1.1 + 0.4 + 1.2 + -0.8 + -3.1 + -1.9
original_mean = original_total / 7
print(math.isclose(mean_celsius, original_mean))   # True
```

## Exercise 8: Round-trip through a file, with an append

```python
from pathlib import Path
import math

path = Path("_files/my_week.txt")
path.parent.mkdir(exist_ok=True)

# 1. "w" creates the file (or overwrites it); write the five values one at a time
with path.open("w", encoding="utf-8") as fhandler:
    fhandler.write("-2.3\n")
    fhandler.write("-1.1\n")
    fhandler.write("0.4\n")
    fhandler.write("1.2\n")
    fhandler.write("-0.8\n")

# 2. "a" appends, keeping what is already there
with path.open("a", encoding="utf-8") as fhandler:
    fhandler.write("-3.1\n")
    fhandler.write("-1.9\n")

# 3. read the seven lines back and cast each one by name
with path.open("r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

print("number of lines:", len(lines))                  # 7
first = float(lines[0])
second = float(lines[1])
third = float(lines[2])
fourth = float(lines[3])
fifth = float(lines[4])
sixth = float(lines[5])
seventh = float(lines[6])
mean_celsius = (first + second + third + fourth + fifth + sixth + seventh) / len(lines)
print(f"mean = {mean_celsius:.2f} °C")                 # mean = -1.09 °C

# 4. the round trip should not change the mean
original_mean = (-2.3 + -1.1 + 0.4 + 1.2 + -0.8 + -3.1 + -1.9) / 7
print(math.isclose(mean_celsius, original_mean))        # True
```

## Exercise 9: A season of station data

The exercises so far used values typed by hand. This one uses a real record: daily maximum
temperatures from a weather station, one observation per day. Work through the steps in order —
each builds on the last, and together they take you from a raw file on disk to a filtered,
unit-converted file of your own.

The data has three columns: the station code, the day as `dd.mm.yy`, and the daily maximum
temperature **in degrees Fahrenheit**. Converting to SI units is part of the job.

```python
# Pre-supplied: download the data file and cache it locally.
import pooch

data_file = pooch.retrieve(
    url="https://raw.githubusercontent.com/gse-unil/2026_MLEES_book/main/data/part-I/Ch1-Lab01-Ex8.csv",
    known_hash="sha256:034755fb289b4e157e5d029995481786a359eb25a80df62c09ee83d063013144",
    fname="ch1-lab01-ex8.csv",
    path=pooch.os_cache("mlees"),
)
print("data cached at:", data_file)
```

**Step 1.** Read the file and look at what actually came off disk.

```python
with open(data_file, "r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

print("lines in file:", len(lines))
for line in lines[:3]:
    print(repr(line))    # repr shows the trailing newline, which print would hide
```

**Step 2.** Split the first record and check what types the fields really are.

```python
first_record = lines[1].strip()      # line 0 is the header
fields = first_record.split(",")
print(fields)

for field in fields:
    print(field, "->", type(field))

# No: every field is a str at this point. Reading a file gives text, and the
# temperature only becomes a number once we cast it ourselves.
```

**Step 3.** Turn that record into three typed, well-named variables.

```python
station_code = fields[0]
day = fields[1]
max_temp_fahrenheit = float(fields[2])
max_temp_celsius = (max_temp_fahrenheit - 32.0) * 5.0 / 9.0

print(f"station {station_code} on {day}: {max_temp_celsius:.1f} °C")
print(type(station_code), type(day), type(max_temp_celsius))
```

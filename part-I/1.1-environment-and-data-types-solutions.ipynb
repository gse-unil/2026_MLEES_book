## Exercise 3: Operator families and a quality flag

```python
reading_1_celsius = 18.2
reading_2_celsius = -300.0
reading_3_celsius = None

# 1. `and` requires both conditions to hold
is_plausible = reading_1_celsius > -273.15 and reading_1_celsius < 60.0
print(is_plausible)                                    # True

# 2. `not` negates a boolean
is_implausible = not (reading_2_celsius > -273.15)
print(is_implausible)                                  # True

# 3. `is` tests identity — the right way to check for None
print(reading_3_celsius is None)                       # True

# 4. a bool casts to int: True -> 1, False -> 0
n_valid = 0
n_valid += int(reading_1_celsius > -273.15)
n_valid += int(reading_2_celsius > -273.15)
print("n_valid:", n_valid)                             # 1
```

## Exercise 5: Indexing and slicing a record

```python
record = "ALO01.03.2022"

# 1. slicing: each field sits at a known position
station = record[:3]        # first three characters
day = record[3:5]
month = record[6:8]
year = record[-4:]          # the last four characters
print(station, day, month, year)      # ALO 01 03 2022

# 2. the same month, via a string method
date_part = record[3:]
month_again = date_part.split(".")[1]
print(month_again)                     # 03

# 3. slicing assumes fixed positions, so a two- or four-character station code shifts every
#    field and silently returns the wrong text. Splitting on a separator survives that,
#    so it is the safer choice whenever the field widths can vary.
```

## Exercise 6: Formatting π to the precision you choose

```python
import math
from pathlib import Path

path = Path("_files/pi_exercise.txt")
path.parent.mkdir(exist_ok=True)

# 1. write the first line with mode "w", which creates the file or truncates it
with path.open("w", encoding="utf-8") as fhandler:
    fhandler.write("This is my first I/O exercise.\n")

# 2. reopen with mode "a", which appends instead of truncating
with path.open("a", encoding="utf-8") as fhandler:
    fhandler.write(f"Pi to four decimals is {math.pi:.4f}.\n")
    fhandler.write("The first line was written in mode 'w', the rest in mode 'a'.\n")
    fhandler.write("Opening in 'w' a second time would have thrown the first line away.\n")

# 3. read it back
with path.open("r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

print(len(lines), "lines:")
print(lines[0].strip())
print(lines[1].strip())
print(lines[2].strip())
print(lines[3].strip())

# 4. two decimals instead of four -- 3.14159... rounds to 3.14
print(f"four decimals: {math.pi:.4f}")
print(f"two decimals:  {math.pi:.2f}")
```

## Exercise 7: Round-trip through a file, with an append

```python
from pathlib import Path

path = Path("_files/my_week.txt")
path.parent.mkdir(exist_ok=True)

# 1. "w" creates the file (or overwrites it); write the five values one at a time
with path.open("w", encoding="utf-8") as fhandler:
    fhandler.write("-2.3\n")
    fhandler.write("-1.1\n")
    fhandler.write("0.4\n")
    fhandler.write("1.2\n")
    fhandler.write("-0.8\n")

# 2. "a" appends, keeping what is already there
with path.open("a", encoding="utf-8") as fhandler:
    fhandler.write("-3.1\n")
    fhandler.write("-1.9\n")

# 3. read the seven lines back and cast each one by name
with path.open("r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

print("number of lines:", len(lines))                  # 7
first = float(lines[0])
second = float(lines[1])
third = float(lines[2])
fourth = float(lines[3])
fifth = float(lines[4])
sixth = float(lines[5])
seventh = float(lines[6])
mean_celsius = (first + second + third + fourth + fifth + sixth + seventh) / len(lines)
print(f"mean = {mean_celsius:.2f} °C")                 # mean = -1.09 °C
```

## Going deeper (optional)

### Exercise 8: Did the round trip change the number?

```python
import math

# 1. the mean read back from the file (from Exercise 7) and the mean computed directly
direct_mean = (-2.3 + -1.1 + 0.4 + 1.2 + -0.8 + -3.1 + -1.9) / 7
print(mean_celsius, direct_mean)

# 2. exact equality
print(mean_celsius == direct_mean)

# 3. equality within a tolerance
print(math.isclose(mean_celsius, direct_mean))

# 4. `==` asks whether two floats have identical bit patterns, which depends on the order the
#    additions happened in and on how many decimals the text file kept. `math.isclose` asks
#    whether they agree to within a tolerance, which is the question you actually mean. Use
#    math.isclose in the script.
```

## Exercise 9: A year of station data

The exercises so far used values typed by hand. This one uses a real record: a full year of daily
maximum temperatures from weather station IIB, one observation per day. Work through the steps in
order — each builds on the last, and together they take you from a raw file on disk to a
unit-converted file of your own.

The data has three columns: the station code, the day as `dd.mm.yy`, and the daily maximum
temperature **in degrees Fahrenheit**. Converting to SI units is part of the job.

```python
# Pre-supplied: download the data file and cache it locally.
# You do not need to understand this cell yet — fetching data is covered in the
# reproducible-data-pipelines bonus subchapter.
import pooch

data_file = pooch.retrieve(
    url="https://raw.githubusercontent.com/gse-unil/2026_MLEES_book/main/data/part-I/station_iib_daily_max_temp_2022.csv",
    known_hash="sha256:034755fb289b4e157e5d029995481786a359eb25a80df62c09ee83d063013144",
    fname="station_iib_daily_max_temp_2022.csv",
    path=pooch.os_cache("mlees"),
)
print("data cached at:", data_file)
```

**Step 1.** Read the file and look at what actually came off disk.

```python
with open(data_file, "r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

print("lines in file:", len(lines))      # 366: one header plus 365 days
print(lines[0].strip())                  # station,day,max_temp_f
print(lines[1].strip())                  # IIB,01.01.22,14.7
print(lines[2].strip())                  # IIB,02.01.22,8.6
```

**Step 2.** Split the first record and check what types the fields really are.

```python
first_record = lines[1].strip()      # line 0 is the header
fields = first_record.split(",")
print(fields)                        # ['IIB', '01.01.22', '14.7']

print(fields[0], "->", type(fields[0]))
print(fields[1], "->", type(fields[1]))
print(fields[2], "->", type(fields[2]))

# No: every field is a str at this point. Reading a file gives text, and the
# temperature only becomes a number once we cast it ourselves.
```

**Step 3.** Turn that record into three typed, well-named variables.

```python
station_code = fields[0]
day_1 = fields[1]
max_temp_fahrenheit_1 = float(fields[2])
max_temp_celsius_1 = (max_temp_fahrenheit_1 - 32.0) * 5.0 / 9.0

print(f"station {station_code} on {day_1}: {max_temp_celsius_1:.1f} °C")
print(type(station_code), type(day_1), type(max_temp_celsius_1))
```

**Step 4.** The same for the second record, then compare the two days.

```python
second_record = lines[2].strip()
fields_2 = second_record.split(",")
day_2 = fields_2[1]
max_temp_fahrenheit_2 = float(fields_2[2])
max_temp_celsius_2 = (max_temp_fahrenheit_2 - 32.0) * 5.0 / 9.0

print(f"station {station_code} on {day_2}: {max_temp_celsius_2:.1f} °C")

# 1. comparing the raw fields, before any casting
print("raw text says day 2 was warmer:", fields_2[2] > fields[2])

# 2. comparing the cast values
print("numbers say day 2 was warmer:", max_temp_celsius_2 > max_temp_celsius_1)

# 3. The numbers are right: 8.6 °F is colder than 14.7 °F, so day 2 was not warmer.
#    The text comparison walks character by character and stops at the first one that
#    differs: '8' comes after '1', so it calls "8.6" the larger string. It never sees
#    that "14.7" has two digits before the decimal point and "8.6" has one.
```

**Step 5.** Write the two converted days out as your own summary file.

```python
from pathlib import Path

summary_path = Path("_files/station_iib_summary.csv")
summary_path.parent.mkdir(exist_ok=True)

with summary_path.open("w", encoding="utf-8") as fhandler:
    fhandler.write("station,day,max_temp_celsius\n")
    fhandler.write(f"{station_code},{day_1},{max_temp_celsius_1:.1f}\n")
    fhandler.write(f"{station_code},{day_2},{max_temp_celsius_2:.1f}\n")

with summary_path.open("r", encoding="utf-8") as fhandler:
    print(fhandler.read())
```

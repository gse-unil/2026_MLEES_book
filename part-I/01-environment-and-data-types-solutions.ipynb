## Exercise 2: Unit conversions

A temperature is recorded as 18.5 °C. Compute it in kelvin and in degrees Fahrenheit. Separately, a discharge of 4.7 m³ s⁻¹ drains a catchment of area 250 km². Express the specific discharge in mm per day (use: 1 m³ s⁻¹ spread over 1 km² for one day equals 86.4 mm).

:::{admonition} Solution
:class: note
```python
temp_celsius = 18.5
temp_kelvin = temp_celsius + 273.15
temp_fahrenheit = temp_celsius * 9 / 5 + 32
print(f"{temp_kelvin:.2f} K, {temp_fahrenheit:.1f} °F")

discharge_m3s = 4.7
area_km2 = 250.0
specific_discharge_mm_day = discharge_m3s / area_km2 * 86.4
print(f"{specific_discharge_mm_day:.3f} mm/day")
```

## Exercise 4: Parse a record

A logger emits records like `"2024-01-15 , JUNGFRAUJOCH , -2.3 , degC"`. Extract the date string, the station name in title case, the numeric value as a float, and the unit in lower case, then print them.

:::{admonition} Solution
:class: note
```python
record = "2024-01-15 , JUNGFRAUJOCH , -2.3 , degC"
fields = record.split(",")
date_str = fields[0].strip()
station = fields[1].strip().title()
value_celsius = float(fields[2].strip())
unit = fields[3].strip().lower()
print(date_str, station, value_celsius, unit)
```

## Exercise 6: Round-trip through a file

Write the readings `[-2.3, -1.1, 0.4, 1.2, -0.8, -3.1, -1.9]` (°C) to `my_week.txt`, one per line, using `pathlib` and a `for` loop. Read them back, cast each line to `float`, and compute the mean (to two decimals) with a `for` loop and `len()`. Then use `math.isclose` to confirm the write-then-read round trip leaves the mean unchanged.

:::{admonition} Solution
:class: note
```python
from pathlib import Path
import math

readings_celsius = [-2.3, -1.1, 0.4, 1.2, -0.8, -3.1, -1.9]
path = Path("my_week.txt")

# write one value per line, building the lines with a loop
lines = []
for value in readings_celsius:
    lines.append(str(value))
path.write_text("\n".join(lines) + "\n", encoding="utf-8")

# read back and cast each line to float
recovered = []
for line in path.read_text(encoding="utf-8").splitlines():
    recovered.append(float(line))

# mean of the recovered values, using a loop and len()
total = 0.0
for value in recovered:
    total += value
mean_celsius = total / len(recovered)
print(f"mean = {mean_celsius:.2f} °C")

# the round trip should not change the mean
original_total = 0.0
for value in readings_celsius:
    original_total += value
print(math.isclose(mean_celsius, original_total / len(readings_celsius)))
```

## Exercise 7: Never trust `==` on floats

Show that `0.1 + 0.2 == 0.3` is `False`, print the actual value of `0.1 + 0.2`, and then write a correct equality check using `math.isclose`.

:::{admonition} Solution
:class: note
```python
import math
print(0.1 + 0.2 == 0.3)              # False
print(0.1 + 0.2)                     # 0.30000000000000004
print(math.isclose(0.1 + 0.2, 0.3))  # True
```
